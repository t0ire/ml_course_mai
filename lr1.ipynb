# Анализ ошибок

лучшая модель - randon forest.

тк RMSE — это средняя ошибка то она может скрывать: объекты с очень большими ошибками. группы объектов где модель систематически ошибается. значения quality которые модель предсказывает плохо

```python
import matplotlib.pyplot as plt

errors_rf = y_w_test - y_w_pred_forest

print(f"средняя ошибка:      {errors_rf.mean():.4f}")
print(f"стандартное отклонение: {errors_rf.std():.4f}")
print(f"минимальная ошибка:  {errors_rf.min():.4f}")
print(f"максимальная ошибка: {errors_rf.max():.4f}")
print(f"медиана ошибки:      {errors_rf.median():.4f}")

fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(errors_rf, bins=40, color='steelblue', edgecolor='black')
ax.axvline(x=0, color='red', linestyle='--', linewidth=2, label='идеальное предсказание')
ax.axvline(x=errors_rf.mean(), color='orange', linestyle='--',
           linewidth=2, label=f'среднее = {errors_rf.mean():.2f}')
ax.set_xlabel('y_true - y_pred')
ax.set_ylabel('частота')
ax.set_title('распределение ошибок у Random Forest white')
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

некоторые объекты имеют ошибки ±2–3 - это выбросы

```python
errors_df = pd.DataFrame({
    'true': y_w_test.values,
    'pred': y_w_pred_forest,
    'error': errors_rf.values,
    'abs_error': np.abs(errors_rf.values)
})

errors_df['quality_group'] = errors_df['true']

group_stats = errors_df.groupby('quality_group').agg(
    count=('error', 'count'),
    rmse=('error', lambda x: np.sqrt((x**2).mean())),
    mae=('error', lambda x: np.abs(x).mean()),
    mean_error=('error', 'mean')
).reset_index()

group_stats
```


дисбаланс классов:
- quality = 5, 6: 723 объекта
- quality = 3, 8: 40 объектов

чем реже quality тем хуже RMSE:
- quality = 6: RMSE = 0.395 - лучший
- quality = 3: RMSE = 2.755 - худший

модель приближает предсказания к среднему:
- quality = 3, 4: mean_error < 0 - модель завышает
- quality = 7, 8: mean_error > 0 - модель занижает
- quality = 6: mean_error = 0 -  модель не смещена

модель пробивает порог только для типичных quality 5, 6. для редких — нет. это происходит из-за нехватки данных:
- quality = 5, 6: RMSE < 0.70
- quality = 3, 4, 7, 8: RMSE > 0.70

```python
top5 = errors_df.sort_values(by='abs_error', ascending=False).head(5)

top5_indices = top5.index
top5_features = x_w_test.iloc[top5_indices]

top5_full = pd.concat([top5_features.reset_index(drop=True),
                       top5[['true', 'pred', 'error']].reset_index(drop=True)], axis=1)

top5_full
```

модель систематически завышает предсказания для quality = 3 потому что:
- мало примеров в обучающих данных
- усреднение приближает к среднему
- признаки этих объектов нетипичны

```python
errors_gb_r = y_r_test - y_r_pred_gbr

print(f"средняя ошибка:         {errors_gb_r.mean():.4f}")
print(f"стандартное отклонение: {errors_gb_r.std():.4f}")
print(f"минимальная ошибка:     {errors_gb_r.min():.4f}")
print(f"максимальная ошибка:    {errors_gb_r.max():.4f}")
print(f"медиана ошибки:         {errors_gb_r.median():.4f}")

fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(errors_gb_r, bins=40, color='darkorange', edgecolor='black')
ax.axvline(x=0, color='red', linestyle='--', linewidth=2, label='идеальное предсказание')
ax.axvline(x=errors_gb_r.mean(), color='orange', linestyle='--',
           linewidth=2, label=f'среднее = {errors_gb_r.mean():.2f}')
ax.set_xlabel('y_true - y_pred')
ax.set_ylabel('частота')
ax.set_title('распределение ошибок Gradient Boosting red')
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
